## 13. Create the first static Plotly figure

Build the first figure before placing it in Dash. Give it a descriptive title and human-readable axis and legend labels.



```python
city_mpg_by_make = (analysis.groupby("make", as_index=False)["city_mpg"].mean())

city_mpg_figure = px.bar(
    city_mpg_by_make,
    x="make",
    y="city_mpg",
    title="Average City Fuel Economy by Manufacturer",
    labels={
        "make": "Vehicle Manufacturer",
        "city_mpg": "Average City MPG"
    }
)

city_mpg_figure.show()

```

**What evidence should a viewer obtain from the first figure?**
The first figure shows the average city fuel economy for each vehicle manufacturer. A viewer can use it to compare how fuel-efficient vehicles from different manufacturers are during city driving.


## 14. Create the second static Plotly figure

The second figure should add different but related evidence rather than repeat the first figure in another style.



```python
highway_mpg_by_make = (analysis.groupby("make", as_index=False)["highway_mpg"].mean())

highway_mpg_figure = px.bar(
    highway_mpg_by_make,
    x="make",
    y="highway_mpg",
    title="Average Highway Fuel Economy by Manufacturer",
    labels={
        "make": "Vehicle Manufacturer",
        "highway_mpg": "Average Highway MPG"
    }
)

highway_mpg_figure.show()

```

## 16. Create reusable figure code

Write a function that accepts the selected control value, filters or summarizes `analysis`, creates both Plotly figures, and returns the figures in a consistent order. Fix axis ranges or category orders when they need to remain stable across selections.



```python
def update_figures(selected_fuel_type):
    filtered_data = analysis[analysis["fuel_type"] == selected_fuel_type]

    city_mpg_by_make = (
        filtered_data.groupby("make", as_index=False)["city_mpg"]
        .mean()
        .sort_values("make")
    )

    highway_mpg_by_make = (
        filtered_data.groupby("make", as_index=False)["highway_mpg"]
        .mean()
        .sort_values("make")
    )

    manufacturer_order = sorted(analysis["make"].unique())

    city_mpg_figure = px.bar(
        city_mpg_by_make,
        x="make",
        y="city_mpg",
        title=f"Average City Fuel Economy by Manufacturer - {selected_fuel_type.title()}",
        labels={
            "make": "Vehicle Manufacturer",
            "city_mpg": "Average City MPG"
        },
        category_orders={"make": manufacturer_order}
    )

    city_mpg_figure.update_yaxes(range=[0, 130])

    highway_mpg_figure = px.bar(
        highway_mpg_by_make,
        x="make",
        y="highway_mpg",
        title=f"Average Highway Fuel Economy by Manufacturer - {selected_fuel_type.title()}",
        labels={
            "make": "Vehicle Manufacturer",
            "highway_mpg": "Average Highway MPG"
        },
        category_orders={"make": manufacturer_order}
    )

    highway_mpg_figure.update_yaxes(range=[0, 130])

    return city_mpg_figure, highway_mpg_figure

```

## 17. Build the layout

Create a Dash app and its layout. Include the title, purpose statement, interactive control, two empty `dcc.Graph` components with unique IDs, and short conclusion and limitation statements.



```python
app = Dash(__name__)

app.layout = html.Div([
    html.H1("How Does Fuel Type Relate to Fuel Economy?"),

    html.P(
        "This dashboard examines how fuel economy varies across vehicle manufacturers for different fuel types using city and highway MPG."
    ),

    html.Label("Select Fuel Type:"),
    dcc.Dropdown(
        id="fuel-type-dropdown",
        options=fuel_type_options,
        value="gas",
        clearable=False
    ),

    dcc.Graph(
        id="city-mpg-graph"
    ),

    dcc.Graph(
        id="highway-mpg-graph"
    ),

    html.H3("Conclusion"),
    html.P(
        "The dashboard allows viewers to compare city and highway fuel economy across manufacturers for each fuel type."
    ),

    html.H3("Limitation"),
    html.P(
        "The fuel-type groups are highly unbalanced. Gas vehicles account for 546 observations, while diesel and electricity each have only 2 observations, so comparisons involving those fuel types should "
        "be interpreted cautiously."
    )
])

```

## 18. Connect the callback

Connect the control's `value` to the `figure` property of both graphs. The callback function should call the reusable figure function and return its two figures in the same order as the two outputs.



```python
@app.callback(
    Output("city-mpg-graph", "figure"),
    Output("highway-mpg-graph", "figure"),
    Input("fuel-type-dropdown", "value")
)
def update_dashboard(selected_fuel_type):
    return update_figures(selected_fuel_type)

```

## 19. Check the callback as a Python function

Call the callback or reusable figure function with at least two valid control values. Confirm that each call returns two Plotly figures without an error.



```python
gas_city_fig, gas_highway_fig = update_figures("gas")
diesel_city_fig, diesel_highway_fig = update_figures("diesel")

print(type(gas_city_fig))
print(type(gas_highway_fig))

print(type(diesel_city_fig))
print(type(diesel_highway_fig))
```
